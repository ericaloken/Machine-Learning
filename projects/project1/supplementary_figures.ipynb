## model_selection_cv_mse_vs_degree_k5

<img src="figures/pngs/model_selection_cv_mse_vs_degree_k5.png" width="600">


5-fold CV MSE vs degree for OLS, Ridge and Lasso on the training part ($n=400$; 100 points held out as test set), using the best $\lambda$ at each degree. All methods agree at low degree. OLS and Ridge are flat from about degree 12, with OLS rising only slightly above degree 20. Our Lasso falls behind Scikit-Learn for $d \geq 8$.

## ols_bias_variance_vs_degree_both

<img src="figures/pngs/ols_bias_variance_vs_degree_both.png" width="600">


OLS bootstrap bias–variance vs degree ($n=500$), with bias$^2$ measured against both the true $f(x)$ and the noisy $y$. True bias$^2$ falls to about $5\cdot 10^{-4}$ near degree 18 and rises sharply from degree 24, while the variance grows with degree. Measured against $y$, bias$^2$ includes the noise, so test error $=$ bias$^2_y$ $+$ variance exactly. The test error levels off near $\sigma^2$ and rises again from about degree 23.

## ols_bias_variance_vs_degree_different_n_bootstrap

OLS bootstrap MSE, bias$^2$ and variance vs degree for $n=50, 100, 500$. With few points the fits become unstable at high degree (up to $10^{20}$ for $n=50$). $n=500$ stays well behaved up to about degree 25.

## ols_mse_r2_vs_degree_different_sigma

<img src="figures/pngs/ols_mse_r2_vs_degree_different_sigma.png" width="600">


OLS test MSE and $R^2$ vs degree for $\sigma = 0, 0.1, 0.2$. The noise sets the error floor. With $\sigma=0$ the MSE goes to 0 and $R^2$ to 1. With $\sigma=0.2$ the model starts overfitting after about degree 10.

## ols_noise_floor_vs_degree

<img src="figures/pngs/ols_noise_floor_vs_degree.png" width="600">


Noise variance estimated from the bootstrap decomposition, as MSE $-$ bias$^2_f$ $-$ variance and as bias$^2_y$ $-$ bias$^2_f$. The two are identical, since test error $=$ bias$^2_y$ $+$ variance. The estimate lies between 0.012 and 0.016, above $\sigma^2=0.01$, and varies with degree: on a finite test set the noise does not average out, and the cross term between noise and bias is not exactly zero.

## ols_parameters_vs_degree

<img src="figures/pngs/ols_parameters_vs_degree.png" width="600">


OLS coefficients vs degree. Their magnitude grows quickly, to tens at $d=10$ and thousands at $d=15$, with alternating signs. This is a sign of ill-conditioning and overfitting.

OLS SGD ($M=10$): constant learning rate (solid) vs inverse time decay (dashed). The decay smooths the curves and helps momentum the most, but slows the other optimizers. None of them reach $10^{-3}$ within 1000 epochs.

## ridge_shrinkage_factors_degree_15

<img src="figures/pngs/ridge_shrinkage_factors_degree_15.png" width="600">


Ridge shrinkage factors $\sigma_i^2/(\sigma_i^2+\lambda)$ for the degree-15 singular values. Directions with $\sigma_i \ll \sqrt{\lambda}$ are suppressed, and larger $\lambda$ moves the cut-off up. Many singular values are tiny, so even $\lambda=10^{-4}$ removes several directions.

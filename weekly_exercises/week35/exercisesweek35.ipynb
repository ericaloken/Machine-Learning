## Exercise 1: Analytical exercises

In this exercise we derive the expressions for various derivatives of products of
vectors and matrices. Such derivatives are central to the optimization of various
cost functions. Although we will often use automatic differentiation in actual
calculations, having the analytical expressions is extremely helpful in case we
have simpler derivatives, as well as when we analyze various properties (like
second derivatives) of the chosen cost functions. Vectors are always written as
boldfaced lower-case letters and matrices as upper-case boldfaced letters.

Show that

$$
1. \quad \frac{\partial (\boldsymbol{a}^T\boldsymbol{x})}{\partial \boldsymbol{x}} = \boldsymbol{a}^T,
$$

and

$$
2. \quad \frac{\partial (\boldsymbol{a}^T\boldsymbol{A}\boldsymbol{a})}{\partial \boldsymbol{a}} = \boldsymbol{a}^T\left(\boldsymbol{A}+\boldsymbol{A}^T\right),
$$

and

$$
3. \quad \frac{\partial \left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)^T\left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)}{\partial \boldsymbol{s}} = -2\left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)^T\boldsymbol{A},
$$

4. and finally find the second derivative of this function with respect to the vector
$\boldsymbol{s}$.

If we replace the vector $\boldsymbol{s}$ with the unknown parameters
$\boldsymbol{\theta}$ used to define the ordinary least squares method, we end up
with the equations that determine these parameters. The matrix $\boldsymbol{A}$ is
then the design matrix $\boldsymbol{X}$, and $\boldsymbol{x}$ has to be replaced
with the outputs $\boldsymbol{y}$. The second derivative of the mean squared error
is then proportional to the Hessian matrix
$\boldsymbol{H}=\boldsymbol{X}^T\boldsymbol{X}$.

**Hint:** it is always useful to write out with summation indices the various
quantities; take a look at the examples in the week 35 slides. As an example,
consider the function

$$
f(\boldsymbol{x}) = \boldsymbol{A}\boldsymbol{x},
$$

which reads for a specific component $f_i$ (defining $\boldsymbol{A}$ to be an
$n\times n$ matrix and $\boldsymbol{x}$ a vector of length $n$)

$$
f_i = \sum_{j=0}^{n-1} a_{ij} x_j,
$$

which leads to

$$
\frac{\partial f_i}{\partial x_j} = a_{ij},
$$

and written out in terms of the vector $\boldsymbol{x}$ we have

$$
\frac{\partial f(\boldsymbol{x})}{\partial \boldsymbol{x}} = \boldsymbol{A}.
$$


**1.**

$$
\partial_{x_k}(a_j x_j) = a_j \delta_{jk} = a_k
$$

Giving:

$$
\frac{\partial (\boldsymbol{a}^T\boldsymbol{x})}{\partial \boldsymbol{x}} = \boldsymbol{a}^T
$$

**2.**

$$
\partial_{a_k}(a_i A_{ij} a_j) = \delta_{ik} A_{ij} a_j + a_i A_{ij} \delta_{jk} = A_{kj}a_j + a_i A_{ik}
$$

So we have:

$$
\partial_{a_k}(\boldsymbol{a}^T\boldsymbol{A}\boldsymbol{a}) = (\boldsymbol{Aa})_k + (\boldsymbol{A}^T\boldsymbol{a})_k
$$

Giving:

$$
\frac{\partial (\boldsymbol{a}^T\boldsymbol{A}\boldsymbol{a})}{\partial \boldsymbol{a}} = \boldsymbol{a}^T\boldsymbol{A}^T + \boldsymbol{a}^T\boldsymbol{A} = \boldsymbol{a}^T\left(\boldsymbol{A}+\boldsymbol{A}^T\right)
$$

**3.**

Let

$$
r_i = x_i - A_{ij}s_j
$$

Then

$$
\partial_{s_k}(r_i r_i) = \partial_{s_k}\left[(x_i - A_{ij}s_j)(x_i - A_{il}s_l)\right] = \partial_{s_k}\left(x_ix_i - 2x_iA_{ij}s_j + A_{ij}s_jA_{il}s_l\right)
$$

$$
= -2x_iA_{ik} + 2A_{ij}s_jA_{ik} = -2A_{ik}(x_i - A_{ij}s_j) = -2r_iA_{ik}
$$

Giving:

$$
\frac{\partial \left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)^T\left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)}{\partial \boldsymbol{s}} = -2\left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)^T\boldsymbol{A}
$$

**4. Second derivative**

From part 3:

$$
\partial_{s_k} f = -2A_{ik}x_i + 2A_{ik}A_{ij}s_j
$$

Differentiate with respect to $s_l$, we have:

$$
\partial_{s_l}\partial_{s_k} f = 2A_{ik}A_{il} = 2(\boldsymbol{A}^T\boldsymbol{A})_{kl}
$$

so as a matrix,

$$
\frac{\partial^2 f}{\partial \boldsymbol{s}\,\partial \boldsymbol{s}^T} = 2\boldsymbol{A}^T\boldsymbol{A}
$$

This matrix is symmetric, and positive semi-definite since for any vector $\boldsymbol{v}$,

$$
\boldsymbol{v}^T(\boldsymbol{A}^T\boldsymbol{A})\boldsymbol{v} = (\boldsymbol{Av})^T(\boldsymbol{Av}) = \|\boldsymbol{Av}\|^2 \geq 0
$$

so $f$ is convex in $\boldsymbol{s}$. 
Interchaning $\bold{A}$ with the design matrix $\bold{X}$ the second derivative can be written in therms of the Hassian matrix

$$
\frac{\partial^2 f}{\partial \boldsymbol{s}\,\partial \boldsymbol{s}^T} = 2\bold{H}
$$

## Exercise 2: Making your own data and exploring scikit-learn

We will generate our own dataset for a function $y(x)$ where $x\in[0,1]$ and
defined by random numbers computed with the uniform distribution. The function $y$
is a quadratic polynomial in $x$ with added stochastic noise according to the

define our $x$ and $y$ values (with 100 data points).

```python
import numpy as np

rng = np.random.default_rng(2026)
x = rng.random((100, 1))
y = 2.0 + 5.0 * x * x + 0.1 * rng.standard_normal((100, 1))
```

3. Using scikit-learn, compute also the mean squared error, a risk metric
   corresponding to the expected value of the squared (quadratic) error

$$
MSE(\boldsymbol{y},\tilde{\boldsymbol{y}}) = \frac{1}{n}\sum_{i=0}^{n-1}\left(y_i-\tilde{y}_i\right)^2,
$$

   and the $R^2$ score function. If $\tilde{y}_i$ is the predicted value of the
   $i$-th sample and $y_i$ is the corresponding true value, then the $R^2$ score is

$$
R^2(\boldsymbol{y},\tilde{\boldsymbol{y}}) = 1 - \frac{\sum_{i=0}^{n-1}(y_i-\tilde{y}_i)^2}{\sum_{i=0}^{n-1}(y_i-\bar{y})^2},
$$

   where the mean value of $\boldsymbol{y}$ is

$$
\bar{y} = \frac{1}{n}\sum_{i=0}^{n-1} y_i.
$$

   You can use the functionality included in scikit-learn, or your own functions
   from the Tuesday notebook. Discuss the meaning of these results. Try also to
   vary the coefficient in front of the added stochastic noise term and discuss
   the quality of the fits.

```python
# Your code for exercise 2 here
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

def poly_of_degree(x, y, degree):
    x_train, x_test, y_train, y_test = train_test_split(
        x, y, test_size=0.2, random_state=42
    )
    N = degree + 1
    X_train = np.zeros ((len(x_train), N))
    X_test = np.zeros ((len(x_test), N))

    for j in range(N):
        X_train[:, j] = x_train[:, 0] ** j
        X_test[:, j] = x_test[:, 0] ** j

    theta = np.linalg.inv(X_train.T @ X_train) @ X_train.T @ y_train

    return x_train, x_test, y_train, y_test, X_train, X_test, theta

def ols(X, y):
    U, s, Vt = np.linalg.svd(X, full_matrices=False)
    theta = Vt.T @ (U.T @ y / s.reshape(-1, 1))
    #print(f"s:{s}")
    return theta

def mse(y, y_pred):
    N = len(y)
    mse_manual = np.sum((y - y_pred) ** 2) / N
    mse_sklearn = mean_squared_error(y, y_pred)

    return mse_manual, mse_sklearn

def r2(y, y_pred):
    r2_manual = 1 - np.sum((y - y_pred) ** 2) / np.sum((y - np.mean(y)) ** 2)
    r2_sklearn = r2_score(y, y_pred)
    return r2_manual, r2_sklearn


noises = [0.0, 0.1, 0.5, 1.0, 2.0]
degree = 2

for noise in noises:
    rng = np.random.default_rng(2026)
    x = rng.random((100, 1))
    y = 2.0 + 5.0 * x * x + noise * rng.standard_normal((100, 1))

    x_train, x_test, y_train, y_test, X_train, X_test, theta = poly_of_degree(x, y, degree)

    # manual OLS using SVD
    theta_manual = ols(X_train, y_train)
    y_train_pred_manual = X_train @ theta_manual
    y_test_pred_manual = X_test @ theta_manual

    # Scikit-learn OLS
    model = LinearRegression(fit_intercept=False)
    model.fit(X_train, y_train)
    theta_sklearn = model.coef_.ravel()

    y_train_pred_sklearn = model.predict(X_train)
    y_test_pred_sklearn = model.predict(X_test)

    # compute MSE, R2
    mse_train_manual, mse_train_sklearn = mse(y_train, y_train_pred_manual)
    mse_test_manual, mse_test_sklearn = mse(y_test, y_test_pred_manual)

    r2_train_manual, r2_train_sklearn = r2(y_train, y_train_pred_manual)
    r2_test_manual, r2_test_sklearn = r2(y_test, y_test_pred_manual)

    print(f"\nNoise = {noise}")
    print("----------------------------------------")

    print("Manual theta:       ", theta_manual.ravel())
    print("Scikit-learn theta: ", theta_sklearn)

    print(
        f"Train MSE = {mse_train_manual:.4f}, "
        f"Train R^2 = {r2_train_manual:.4f}"
    )

    print(
        f"Test MSE  = {mse_test_manual:.4f}, "
        f"Test R^2  = {r2_test_manual:.4f}"
    )


```

EXPLANATION:

We see from the print that with noise = 0 we recover the true coefficients $(2,0,5)$ almost perfectly, and MSE $\approx$ 0 and $R^2 \approx 1$, indicating a perfect fit. As the noise increases the fitted parameters deviate more from the true coefficients, so the MSE increases while $R^2$ decreases. This is expected because the polynomial model can predict the true function but not the stochastic noise. $R^2$ increasing indicates that a smaller fraction of the variance is explained by the model as the noise increases. 

## Exercise 3: Split data in test and training data

In this exercise we want you to compute the MSE for the training data and the test
data as a function of the complexity of a polynomial, that is the degree of a given
polynomial.

The aim is to reproduce Figure 2.11 of Hastie et al., *The Elements of Statistical
Learning*. Feel free to read the discussions leading to that figure.

Our data is defined by $x\in[-3,3]$ with a total of, for example, $n=100$ data
points. You should try to vary the number of data points $n$ in your analysis.

```python
rng = np.random.default_rng(2026)
n = 100
x = np.linspace(-3, 3, n).reshape(-1, 1)
y = np.exp(-x**2) + 1.5 * np.exp(-(x - 2)**2) + rng.normal(0, 0.1, x.shape)
```

where $y$ is the function we want to fit with a given polynomial.

1. Write a first code which sets up a design matrix $\boldsymbol{X}$ defined by a
   fifth-order polynomial and split your data set into training and test data
   (`sklearn.model_selection.train_test_split` is fine).

2. Write thereafter (using either scikit-learn or your matrix inversion/SVD code)
   an ordinary least squares fit and compute the mean squared error for the
   training data and the test data. These calculations should apply to a model
   given by a fifth-order polynomial. If you compare your own code with
   scikit-learn, note that the latter does not include the intercept by default —

3. Add now a model which allows you to make polynomials up to degree $15$. Perform
   a standard OLS fit of the training data and compute the MSE for the training
   and test data and plot both as functions of the polynomial degree. Compare what
   you see with Figure 2.11 of Hastie et al. Comment on your results. For which
   polynomial degree do you find the optimal (smallest) test MSE?

```python
# Your code for exercise 3 here
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

max_degree = 20
degrees = np.arange(1, max_degree + 1)

mse_train_list = np.zeros(max_degree)
mse_test_list = np.zeros(max_degree)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

for i, degree in enumerate(degrees):
    Xtr = np.vander(x_train.flatten(), degree + 1, increasing=True)
    Xte = np.vander(x_test.flatten(), degree + 1, increasing=True)
    theta = np.linalg.lstsq(Xtr, y_train, rcond=None)[0]  # least square

    y_train_pred = Xtr @ theta
    y_test_pred = Xte @ theta

    mse_train = mean_squared_error(y_train, y_train_pred)
    mse_test = mean_squared_error(y_test, y_test_pred)

    mse_train_list[i] = mse_train
    mse_test_list[i] = mse_test

idx = np.argmin(mse_test_list)
best_deg = degrees[idx]
best_mse = mse_test_list[idx]
print(f"Best test MSE = {best_mse}")
```

```python
from matplotlib import pyplot as plt


plt.figure(figsize=(10, 6))
plt.plot(degrees, mse_train_list, label='Train MSE', marker='o')
plt.plot(degrees, mse_test_list, label='Test MSE', marker='s')
plt.axvline(best_deg, color="gray", linestyle=":", label=f"best test MSE: degree {best_deg}")
plt.axhline(0.01, color="gray", linestyle="--", label=r"noise floor $\sigma^2$ = 0.01")
plt.yscale('log')
plt.xlabel('polynomial degree')
plt.ylabel('MSE')
plt.legend(loc='upper left')
plt.show()
```

EXPLANATION:

From this plot we see that the MSE initially decreases as the polynomial degree increases, with a polynomial degree 10 giving the lowest test MSE,  but this can be affected by the random noise in the data. For higher polynomial degrees, the test MSE icreases while the training MSE continues to decrease. This happens because a more complex model (higher polynomial degree) can adapt more closely to the training data, including the random noise. However, the model predicts new unseen (test) data poorly even though it predicts the training data accurately. We have plotted 20 degrees rather than 15 to better show the expected behaviour.


## Exercise 4 (optional): a first taste of Ridge

This voluntary exercise connects directly to Monday's lecture and to Case 4 of the
Tuesday notebook; it is also a useful warm-up for Project 1.

Keep the data set and the degree-15 design matrix from exercise 3, standardise the
feature columns (training statistics only!), and replace OLS by Ridge regression,

$$
\hat{\boldsymbol{\theta}}_{\mathrm{Ridge}} = \left(\boldsymbol{X}^T\boldsymbol{X}+\lambda\boldsymbol{I}\right)^{-1}\boldsymbol{X}^T\boldsymbol{y}.
$$

1. Plot the training and test MSE as functions of $\lambda$ on a logarithmic grid,
   say $\lambda\in[10^{-8},10^{4}]$. Do you find the U-shape discussed in the
   lecture?

2. Compare your best test MSE with the best value found in exercise 3 by varying
   the polynomial degree. Two dials, one phenomenon — which gives the finer
   control?

3. Check explicitly that your training MSE is monotonically increasing in
   $\lambda$, and explain in one sentence why it must be.

```python
# Your code for exercise 4 here
import matplotlib.pyplot as plt

def ridge_svd(X, y, lmbda):
    U, s, Vt = np.linalg.svd(X, full_matrices=False)
    factor = (s / (s**2 + lmbda)).reshape(-1, 1)
    return Vt.T @ (factor * (U.T @ y))


x_train, x_test, y_train, y_test, Xtr, Xte, theta = poly_of_degree(x, y, 15)

# Work on centred data so no intercept is needed
Xtr_c = Xtr[:, 1:] - Xtr[:, 1:].mean(axis=0)
Xte_c = Xte[:, 1:] - Xte[:, 1:].mean(axis=0)

# centre y using training mean
y_train_c = y_train - y_train.mean()

lmbdas = np.logspace(-8, 4, 100)

mse_train_list = []
mse_test_list = []

for i in range(len(lmbdas)):
    theta_train = ridge_svd(Xtr_c, y_train_c, lmbdas[i])

    # Add intercept back through y_mean
    y_train_pred = y_train.mean() + Xtr_c @ theta_train
    y_test_pred = y_train.mean() + Xte_c @ theta_train

    mse_train, _ = mse(y_train, y_train_pred)
    mse_test, _ = mse(y_test, y_test_pred)

    mse_train_list.append(mse_train)
    mse_test_list.append(mse_test)


best_index = np.argmin(mse_test_list)
best_lambda = lmbdas[best_index]
best_test_mse = mse_test_list[best_index]

plt.figure(figsize=(8, 5))

plt.plot(lmbdas, mse_train_list, 'o-', label="Training MSE")

plt.semilogx(lmbdas, mse_test_list, 'o-', label="Test MSE")

plt.scatter(
    best_lambda,
    best_test_mse,
    color="red",
    zorder=3,
    label=fr"Minimum test MSE: $\lambda={best_lambda:.2e}$"
)

plt.xlabel(r"Regularization parameter $\lambda$")
plt.ylabel("MSE")
plt.title("Degree-15 Ridge regression")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.show()

print(f"Best lambda: {best_lambda:.4e}")
print(f"Best test MSE: {best_test_mse:.6f}")

# check if training MSE increases monotonically
is_monotonic = np.all(np.diff(mse_train_list) >= -1e-12)    # allow small differences due to fp errors
print("Training MSE is monotonically increasing:", is_monotonic)
```

EXPLANATION:

1) We see that the test MSE has a U-shaped behaviour as function of $\lambda$. For very small $\lambda$, Ridge regression is very close to OLS, and the model with a 15-degree polynomial overfits the data. When we increase $\lambda$ the coefficients become smaller and the test MSE eventually decreases because the regularization reduces overfitting. The minimum test MSE occurs at $\lambda \approx 1.3$. For $\lambda$ larger than this, the regularization is too strong and the test MSE increases.

2) In exercise 3 we controlled the complexity of the model by varying the polynomial degree, while in this exercise we vary the regularization with fixed polynomial degree. The best test MSE form exercise 3 was $\sim 0.007$ while with Ridge regression the best test MSE was $\sim 0.008$. Both methods control the same phenomenon, namely overfitting of data, but Ridge regression gives a finer control because $\lambda$ can be varied continuously while the polynomial dgeree can only take discrete integer values. 


3) The training MSE increases with $\lambda$, as expected. In Ridge regression we minimize the training error while also penalizing. This hinders the model from fitting the training data prefectly for large $\lambda$ as the coefficients are getting smaller.

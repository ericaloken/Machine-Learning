## Exercise 1: Analytical exercises — the statistics of the OLS estimator

Assume the data are generated by the linear model of Eq. (2.37),
$\boldsymbol{y} = \boldsymbol{X}\boldsymbol{\theta} + \boldsymbol{\varepsilon}$,
with independent noise $\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$ and a
fixed (non-stochastic) design matrix $\boldsymbol{X}$ of dimension $n \times p$.

**1a)** Show that the expectation value and variance of a single output are
(Eq. (2.39))

$$
\mathbb{E}[y_i] = \boldsymbol{X}_{i,*}\,\boldsymbol{\theta},
\qquad
\mathrm{Var}(y_i) = \sigma^2 .
$$

**1b)** Show that the OLS estimator is unbiased,
$\mathbb{E}[\hat{\boldsymbol{\theta}}] = \boldsymbol{\theta}$ (Eq. (2.40)), and
derive its variance (Eq. (2.42)),

$$
\mathrm{Var}(\hat{\boldsymbol{\theta}}) = \sigma^2\left(\boldsymbol{X}^T\boldsymbol{X}\right)^{-1}.
$$

*Hint:* use
$\mathbb{E}[\boldsymbol{y}\boldsymbol{y}^T] = \boldsymbol{X}\boldsymbol{\theta}\boldsymbol{\theta}^T\boldsymbol{X}^T + \sigma^2\boldsymbol{I}_{nn}$.

**1c)** Optional: Explain how the diagonal elements of
$\sigma^2(\boldsymbol{X}^T\boldsymbol{X})^{-1}$ give a confidence interval for
the coefficient $\theta_j$ (Eq. (2.43)), and why in practice $\sigma^2$ is
replaced by
$\hat{\sigma}^2 = \|\boldsymbol{y}-\boldsymbol{X}\hat{\boldsymbol{\theta}}\|_2^2/(n-p)$.
Where does the $n-p$ come from? (Think of Bessel's correction, Eq. (2.28).)

## Solution exercise 1a)

We assume that the error term $\epsilon$ is normally distributed with mean zero and variance $\sigma^2$, so

$$
\epsilon \sim \mathcal{N}(0,\sigma^2), \qquad p(\epsilon) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{\epsilon^2}{2\sigma^2} \right).
$$

The linear regression model is $\boldsymbol{y} = \boldsymbol{X}\boldsymbol{\theta} + \boldsymbol{\epsilon}$. Using the linearity of expectation,

$$
\mathbb{E}[\boldsymbol{y}] = \mathbb{E}[\boldsymbol{X}\boldsymbol{\theta} + \boldsymbol{\epsilon}]
=
\mathbb{E}[\boldsymbol{X}\boldsymbol{\theta}] + \mathbb{E}[\boldsymbol{\epsilon}].
$$

Since $\boldsymbol{X}$ and $\boldsymbol{\theta}$ are deterministic, $\mathbb{E}[\boldsymbol{X}\boldsymbol{\theta}] = \boldsymbol{X}\boldsymbol{\theta}$. Furthermore, the error term has zero mean,

$$
\mathbb{E}[\epsilon] = \int_{-\infty}^{\infty} \epsilon\,p(\epsilon)\,d\epsilon = 0,
$$

because $\epsilon p(\epsilon)$ is an odd function integrated over the symmetric interval $(-\infty,\infty)$. Therefore,

$$
\mathbb{E}[\boldsymbol{y}] = \boldsymbol{X}\boldsymbol{\theta}.
$$

For a single observation,

$$
y_i = \boldsymbol{X}_{i,*}\boldsymbol{\theta} + \epsilon_i,
$$

where $\boldsymbol{X}_{i,*}$ denotes the $i$-th row of the design matrix. Its expectation value is therefore

$$
\mathbb{E}[y_i] = \boldsymbol{X}_{i,*}\boldsymbol{\theta}.
$$

To determine the variance of $y_i$, we start from the definition

$$
\operatorname{Var}(y_i) = \mathbb{E} \left[ \left( y_i-\mathbb{E}[y_i] \right)^2 \right].
$$

Substituting the expression for $y_i$,

$$
\operatorname{Var}(y_i)
=
\mathbb{E}\left[
\left(
\boldsymbol{X}_{i,*}\boldsymbol{\theta}
+\epsilon_i
-\boldsymbol{X}_{i,*}\boldsymbol{\theta}
\right)^2
\right]
=
\mathbb{E}[\epsilon_i^2].
$$

Because $\mathbb{E}[\epsilon_i]=0$, we have

$$
\operatorname{Var}(\epsilon_i) = \mathbb{E}[\epsilon_i^2] = \sigma^2.
$$

This can also be shown directly from the Gaussian probability density by integration. Dropping the subscript $i$ for simplicity,

$$
\mathbb{E}[\epsilon^2] = \int_{-\infty}^{\infty} \epsilon^2 p(\epsilon)\,d\epsilon,
$$

which gives

$$
\mathbb{E}[\epsilon^2] = \int_{-\infty}^{\infty} \epsilon^2 \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(
-\frac{\epsilon^2}{2\sigma^2} \right) \,d\epsilon.
$$

Introducing the substitution

$$
z=\frac{\epsilon}{\sigma}, \qquad \epsilon=\sigma z, \qquad d\epsilon=\sigma\,dz,
$$

we obtain

$$
\mathbb{E}[\epsilon^2] 
= \frac{\sigma^2}{\sqrt{2\pi}} \int_{-\infty}^{\infty} z^2 e^{-z^2/2}\,dz 
= \frac{\sigma^2}{\sqrt{2\pi}} \sqrt{2\pi} 
= \sigma^2. 
$$

Hence, $\operatorname{Var}(y_i)=\sigma^2$. Thus, each output $y_i$ has mean

$$
\boxed{\mathbb{E}[y_i] = \boldsymbol{X}_{i,*}\boldsymbol{\theta}}
$$

and variance

$$
\boxed{\operatorname{Var}(y_i) = \sigma^2}
$$

## Solution exercise 1b)

We want to show that the OLS estimator is unbiased. The least-squares estimator for $\boldsymbol{\theta}$ is

$$
\hat{\boldsymbol{\theta}} =
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T\boldsymbol{y}.
$$
The expectation value of $\hat{\boldsymbol{\theta}}$ is 
$$
\mathbb{E}[\hat{\boldsymbol{\theta}}] = \mathbb{E}
\left[
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T\boldsymbol{y}
\right] 
=
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T
\mathbb{E}[\boldsymbol{y}],
$$
which is allowed because $(\boldsymbol{X}^T\boldsymbol{X})^{-1}\boldsymbol{X}^T$ does not depend on $\boldsymbol{y}$. From 1a) we have $\mathbb{E}[\boldsymbol{y}] = \boldsymbol{X}\boldsymbol{\theta}$, so

$$
\mathbb{E}[\hat{\boldsymbol{\theta}}]
=
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T
\boldsymbol{X}\boldsymbol{\theta}
= 
\boldsymbol{\theta},
$$
where we have used that $(\boldsymbol{X}^T\boldsymbol{X})^{-1} (\boldsymbol{X}^T\boldsymbol{X}) = \boldsymbol{I_{nn}}$. Thus, we have shown that the OLS estimator is unbiased. 


We now calculate the variance of $\hat{\boldsymbol{\theta}}$,

$$
\operatorname{Var}(\hat{\boldsymbol{\theta}})
=
\mathbb{E} \left[ \hat{\boldsymbol{\theta}} \hat{\boldsymbol{\theta}}^T \right]
-
\mathbb{E}[\hat{\boldsymbol{\theta}}]
\mathbb{E}[\hat{\boldsymbol{\theta}}]^T.
$$

Utilizing that the estimator is unbiases, $\mathbb{E}[\hat{\boldsymbol{\theta}}] = \boldsymbol{\theta}$, and substituting $\hat{\boldsymbol{\theta}} = (\boldsymbol{X}^T\boldsymbol{X})^{-1} \boldsymbol{X}^T\boldsymbol{y}$ yields 

$$
\operatorname{Var}(\hat{\boldsymbol{\theta}})
=
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T
\mathbb{E}
\left[
\boldsymbol{y}\boldsymbol{y}^T
\right]
\boldsymbol{X}
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
-
\boldsymbol{\theta}\boldsymbol{\theta}^T.
$$

The covariance matrix of $\boldsymbol{y}$ is $\operatorname{Var}(\boldsymbol{y}) = \sigma^2\boldsymbol{I_{nn}}$. Using the hint,

$$
\mathbb{E}
\left[
\boldsymbol{y}\boldsymbol{y}^T
\right]
=
\boldsymbol{X}\boldsymbol{\theta}
\boldsymbol{\theta}^T
\boldsymbol{X}^T
+
\sigma^2\boldsymbol{I_{nn}},
$$

we obtain

$$
\operatorname{Var}(\hat{\boldsymbol{\theta}})
=
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T
\left(
\boldsymbol{X}\boldsymbol{\theta}
\boldsymbol{\theta}^T
\boldsymbol{X}^T
+
\sigma^2\boldsymbol{I_{nn}}
\right)
\boldsymbol{X}
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
-
\boldsymbol{\theta}\boldsymbol{\theta}^T.
$$

Expanding the expression gives

$$
\begin{aligned}
\operatorname{Var}(\hat{\boldsymbol{\theta}})
&=
\boldsymbol{\theta}\boldsymbol{\theta}^T
+
\sigma^2
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T\boldsymbol{X}
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
-
\boldsymbol{\theta}\boldsymbol{\theta}^T\\
&=
\sigma^2 (\boldsymbol{X}^T\boldsymbol{X})^{-1},
\end{aligned}
$$
where we have used that the the $\boldsymbol{X}$ matrices become the identity and the first and last term cancels. Therefore,

$$
\boxed{\operatorname{Var}(\hat{\boldsymbol{\theta}}) = \sigma^2 (\boldsymbol{X}^T\boldsymbol{X})^{-1}}
$$



## Solution exercise 1c)

In general, the standard deviation is given by $\sigma = \sqrt{\operatorname{Var}}$. Using the result from exercise 1b),

$$
\operatorname{Var}(\hat{\boldsymbol{\theta}}) = \sigma^2 (\boldsymbol{X}^T\boldsymbol{X})^{-1},
$$

the standard deviation of the $j$-th estimated parameter is therefore

$$
\sigma(\hat{\theta}_j) = \sqrt{\sigma^2 
\left[
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\right]_{jj}}
=
\sigma
\sqrt{
\left[
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\right]_{jj}}.
$$

In practice, the true value of $\sigma^2$ is usually unknown and must be estimated from the residuals. We define the residual vector as

$$
\boldsymbol{r} = \boldsymbol{y} - \boldsymbol{X}\hat{\boldsymbol{\theta}}.
$$

Using $\hat{\boldsymbol{\theta}} = (\boldsymbol{X}^T\boldsymbol{X})^{-1} \boldsymbol{X}^T\boldsymbol{y}$, we obtain

$$
\boldsymbol{r}
=
\boldsymbol{y}
-
\boldsymbol{X}
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T
\boldsymbol{y}.
$$

We can write this as

$$
\boldsymbol{r} = \boldsymbol{M}\boldsymbol{y},
$$

where $\boldsymbol{M} = \boldsymbol{I} - \boldsymbol{X} (\boldsymbol{X}^T\boldsymbol{X})^{-1} \boldsymbol{X}^T$ is the residual projection matrix. Since $\boldsymbol{y} =  \boldsymbol{X}\boldsymbol{\theta} + \boldsymbol{\epsilon}$, we get

$$
\boldsymbol{r}
=
\boldsymbol{M} \left( \boldsymbol{X}\boldsymbol{\theta} + \boldsymbol{\epsilon} \right)
=
\boldsymbol{M}\boldsymbol{X}\boldsymbol{\theta} + \boldsymbol{M}\boldsymbol{\epsilon}.
$$

Since

$$
\boldsymbol{M}\boldsymbol{X}
=
\left[
\boldsymbol{I}
-
\boldsymbol{X}
(\boldsymbol{X}^T\boldsymbol{X})^{-1}
\boldsymbol{X}^T
\right]
\boldsymbol{X}
=
0,
$$

we obtain

$$
\boldsymbol{r} = \boldsymbol{M}\boldsymbol{\epsilon}.
$$

Now consider the squared residual $\boldsymbol{r}^T\boldsymbol{r}$. Since $\boldsymbol{M}$ is symmetric and idempotent, that is, 

$$
\boldsymbol{M}^T=\boldsymbol{M}, \qquad \boldsymbol{M}^2=\boldsymbol{M},
$$

we have

$$
\boldsymbol{r}^T\boldsymbol{r}
=
\boldsymbol{\epsilon}^T
\boldsymbol{M}^T
\boldsymbol{M}
\boldsymbol{\epsilon}
=
\boldsymbol{\epsilon}^T
\boldsymbol{M}
\boldsymbol{\epsilon}.
$$

Taking the expectation value,

$$
\mathbb{E}
\left[
\boldsymbol{r}^T\boldsymbol{r}
\right]
=
\mathbb{E}
\left[
\boldsymbol{\epsilon}^T
\boldsymbol{M}
\boldsymbol{\epsilon}
\right].
$$

Using the identities

$$
\mathbb{E}
\left[
\boldsymbol{\epsilon}^T
\boldsymbol{A}
\boldsymbol{\epsilon}
\right]
=
\operatorname{tr}
\left( 
\boldsymbol{A} \mathbb{E} [\boldsymbol{\epsilon} \boldsymbol{\epsilon}^T
]\right), 
\qquad 
\mathbb{E}
[\boldsymbol{\epsilon} \boldsymbol{\epsilon}^T] = \sigma^2\boldsymbol{I},
$$

we obtain

$$
\mathbb{E}
\left[
\boldsymbol{r}^T\boldsymbol{r}
\right]
=
\sigma^2 \operatorname{tr}(\boldsymbol{M}).
$$

The trace of $\boldsymbol{M}$ is

$$
\operatorname{tr}(\boldsymbol{M})
=
\operatorname{tr}(\boldsymbol{I}) - \operatorname{tr}
\left[
\boldsymbol{X} (\boldsymbol{X}^T\boldsymbol{X})^{-1} \boldsymbol{X}^T
\right].
$$

Using the cyclic property of the trace,

$$
\operatorname{tr}
\left[
\boldsymbol{X} (\boldsymbol{X}^T\boldsymbol{X})^{-1} \boldsymbol{X}^T
\right]
=
\operatorname{tr}
\left[
(\boldsymbol{X}^T\boldsymbol{X})^{-1} \boldsymbol{X}^T\boldsymbol{X}
\right]
=
\operatorname{tr}(\boldsymbol{I}_{pp})
=
p.
$$

Since

$$
\operatorname{tr}(\boldsymbol{I}_{nn}) = n,
$$

we get

$$
\operatorname{tr}(\boldsymbol{M}) = n - p.
$$

Therefore,

$$
\mathbb{E} \left[ \boldsymbol{r}^T\boldsymbol{r} \right] = (n-p)\sigma^2.
$$

This gives the estimator

$$
\hat{\sigma}^2 = \frac{ \boldsymbol{r}^T\boldsymbol{r}}{n-p}
$$

or equivalently,

$$
\hat{\sigma}^2 = \frac{ \left\| \boldsymbol{y} - \boldsymbol{X} \hat{\boldsymbol{\theta}} \right\|_2^2}{n-p},
$$

where $n$ is the number of observations and $p$ is the number of fitted parameters. Finally, replacing the unknown $\sigma$ by its estimate $\hat{\sigma}$, the estimated standard deviation of $\hat{\theta}_j$ becomes

$$
\boxed{
\hat{\sigma}(\hat{\theta}_j) = \hat{\sigma} \sqrt{ \left[ (\boldsymbol{X}^T\boldsymbol{X})^{-1} \right]_{jj}}
}
$$

## Exercise 2: Analytical exercises — deriving the bias-variance tradeoff

Assume now, as in Section 2.10, that the true data are generated by
$\boldsymbol{y} = f(\boldsymbol{x}) + \boldsymbol{\varepsilon}$ with
$\boldsymbol{\varepsilon} \sim \mathcal{N}(0, \sigma^2)$ (Eq. (2.44)), and let
$\tilde{\boldsymbol{y}}$ be the prediction of a model fitted on a randomly
drawn training set.

**2a)** Starting from the expected squared error
$\mathbb{E}[(\boldsymbol{y}-\tilde{\boldsymbol{y}})^2]$ and adding and
subtracting $\mathbb{E}[\tilde{\boldsymbol{y}}]$, derive the decomposition of
Eq. (2.47),

$$
\mathbb{E}\left[(\boldsymbol{y}-\tilde{\boldsymbol{y}})^2\right]
= \frac{1}{n}\sum_i\left(f_i-\mathbb{E}[\tilde{\boldsymbol{y}}]\right)^2
+ \frac{1}{n}\sum_i\left(\tilde{y}_i-\mathbb{E}[\tilde{\boldsymbol{y}}]\right)^2
+ \sigma^2 .
$$

Show explicitly that all three cross terms vanish, stating for each one which
assumption kills it (zero-mean noise, independence of noise and training set,
$f$ non-stochastic).

**2b)** Explain in your own words what the expectation
$\mathbb{E}[\cdot]$ runs over, and what each of the three terms means for the
polynomial-fit example of the lecture.

**2c)** In the numerical estimate (Tuesday's Case 1) the bias is computed
against $y_{\mathrm{test}}$ rather than the unknown $f$. Show that the
measured "bias" then contains $\sigma^2$, so the printed identity
`error >= bias + variance` is an inequality.

## Solution exercise 2a)

Writing the expectation value in terms of a sum,

$$
\mathbb{E}\left[
\left(\boldsymbol{y}-\tilde{\boldsymbol{y}}\right)^2
\right]
=
\frac{1}{n}\sum_{i=1}^{n}
\mathbb{E}\left[
\left(y_i-\tilde{y}_i\right)^2
\right].
$$

Since

$$
y_i = f(x_i) + \epsilon_i,
$$

we have

$$
\mathbb{E}\left[
\left(y_i-\tilde{y}_i\right)^2
\right]
=
\mathbb{E}\left[
\left(f(x_i)+\epsilon_i-\tilde{y}_i\right)^2
\right].
$$

Adding and subtracting $\mathbb{E}[\tilde{y}_i]$ gives

$$
\mathbb{E}\left[
\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
+\mathbb{E}[\tilde{y}_i]-\tilde{y}_i
+\epsilon_i
\right)^2
\right].
$$

Define

$$
A = f(x_i)-\mathbb{E}[\tilde{y}_i],
$$

$$
B = \mathbb{E}[\tilde{y}_i]-\tilde{y}_i,
$$

and

$$
C = \epsilon_i.
$$

Then

$$
\mathbb{E}\left[(A+B+C)^2\right]
=
\mathbb{E}\left[
A^2+B^2+C^2+2AB+2AC+2BC
\right].
$$

The first three terms are

$$
\mathbb{E}[A^2]
=
\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
\right)^2,
$$

$$
\mathbb{E}[B^2]
=
\mathbb{E}\left[
\left(
\tilde{y}_i-\mathbb{E}[\tilde{y}_i]
\right)^2
\right]
=
\operatorname{Var}(\tilde{y}_i),
$$

and

$$
\mathbb{E}[C^2]
=
\mathbb{E}[\epsilon_i^2]
=
\sigma^2.
$$

For the cross terms,

$$
\mathbb{E}[AB]
=
\mathbb{E}\left[
\left(f(x_i)-\mathbb{E}[\tilde{y}_i]\right)
\left(\tilde{y}_i-\mathbb{E}[\tilde{y}_i]\right)
\right].
$$

Since $f(x_i)-\mathbb{E}[\tilde{y}_i]$ is not stochastic, it can
be taken outside the expectation value:

$$
\mathbb{E}[AB]
=
\left(f(x_i)-\mathbb{E}[\tilde{y}_i]\right)
\mathbb{E}\left[
\tilde{y}_i-\mathbb{E}[\tilde{y}_i]
\right]
=0.
$$

Similarly,

$$
\mathbb{E}[AC]
=
\left(f(x_i)-\mathbb{E}[\tilde{y}_i]\right)
\mathbb{E}[\epsilon_i]
=0,
$$

since

$$
\mathbb{E}[\epsilon_i]=0.
$$

Finally,

$$
\mathbb{E}[BC]
=
\mathbb{E}\left[
\left(
\tilde{y}_i-\mathbb{E}[\tilde{y}_i]
\right)\epsilon_i
\right]
=0,
$$

because the noise has zero mean and is assumed to be independent
of the training set.

Therefore,

$$
\boxed{
\mathbb{E}\left[
\left(\boldsymbol{y}-\tilde{\boldsymbol{y}}\right)^2
\right]
=
\frac{1}{n}
\sum_{i=1}^{n}
\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
\right)^2
+
\frac{1}{n}
\sum_{i=1}^{n}
\operatorname{Var}(\tilde{y}_i)
+
\sigma^2
}
$$

or, equivalently,

$$

\text{MSE}
=
\text{Bias}^2
+
\text{Variance}
+
\text{Noise}.

$$


## Solution exercise 2b)

The term

$$
\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
\right)^2
$$

is the squared bias. It describes how well the average prediction
of our model agrees with the true function. A large bias means that the
mean prediction of the model is systematically far away from the true
value.

The term

$$
\operatorname{Var}(\tilde{y}_i)
=
\mathbb{E}\left[
\left(
\tilde{y}_i-\mathbb{E}[\tilde{y}_i]
\right)^2
\right]
$$

is the variance of our model. It tells us how much the fitted model
changes between different training sets, or equivalently how spread out
the predictions are around their mean.

Finally,

$$
\mathbb{E}[\epsilon_i^2] = \sigma^2
$$

is the irreducible noise. This contribution cannot be removed by
changing the model, since it originates from the stochastic noise in the
data.


## Solution exercise 2c)

Suppose that

$$
y_{\text{test},i}
=
f(x_i)+\epsilon_{\text{test},i},
$$

where the test noise satisfies

$$
\mathbb{E}[\epsilon_{\text{test},i}] = 0,
\qquad
\mathbb{E}[\epsilon_{\text{test},i}^2] = \sigma^2.
$$

If we use the test data to estimate the squared bias, we obtain

$$
\mathbb{E}\left[
\left(
y_{\text{test},i}
-
\mathbb{E}[\tilde{y}_i]
\right)^2
\right].
$$

Substituting

$$
y_{\text{test},i}
=
f(x_i)+\epsilon_{\text{test},i},
$$

gives

$$
\mathbb{E}\left[
\left(
f(x_i)
+
\epsilon_{\text{test},i}
-
\mathbb{E}[\tilde{y}_i]
\right)^2
\right].
$$

Expanding,

$$
=
\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
\right)^2
+
2\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
\right)
\mathbb{E}[\epsilon_{\text{test},i}]
+
\mathbb{E}[\epsilon_{\text{test},i}^2].
$$

Since

$$
\mathbb{E}[\epsilon_{\text{test},i}] = 0,
$$

we get

$$
\mathbb{E}\left[
\left(
y_{\text{test},i}
-
\mathbb{E}[\tilde{y}_i]
\right)^2
\right]
=
\left(
f(x_i)-\mathbb{E}[\tilde{y}_i]
\right)^2
+
\sigma^2
.
$$

Thus, the bias measured using the test data contains an additional
contribution from the noise:

$$

\text{measured bias}^2
=
\text{true bias}^2+\sigma^2.

$$

Therefore, if $\sigma^2$ is known, the true squared bias can be estimated as

$$
\text{true bias}^2
=
\text{measured bias}^2-\sigma^2.
$$

Using the bias-variance decomposition,

$$
\mathbb{E}\left[
\left(y_i-\tilde{y}_i\right)^2
\right]
=
\text{Bias}^2
+
\text{Variance}
+
\sigma^2.
$$

Hence the irreducible noise $\sigma^2$ sets the upper limit on the error. As the bias approches the true bias the upper limit of the error

$$
\boxed{
\text{error}
\geq 
\text{bias}^2+\text{variance}.
}
$$

```python
# Your code for exercise 3 here

def bias_variance(n, max_deg=13, n_bootstraps=100):
    """
    This function performs n_boostraps and computes the bias and
    variance for polynomial degree 0 to max_deg.
    """
    np.random.seed(2026)
    x = np.linspace(-3, 3, n).reshape(-1, 1)
    y = np.exp(-x**2) + 1.5 * np.exp(-(x - 2)**2) + np.random.normal(0, 0.1, x.shape)

    degrees = np.arange(max_deg + 1)

    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2026)

    error = np.zeros(max_deg + 1)
    bias = np.zeros(max_deg + 1)
    var = np.zeros(max_deg + 1)

    for degree in degrees:
        model = make_pipeline(
            PolynomialFeatures(degree=degree),
            LinearRegression(fit_intercept=False)
        )

        # one column for each bootstrap
        y_pred = np.zeros((len(y_test), n_bootstraps))

        # start resampling
        for i in range(n_bootstraps):
            x_b, y_b = resample(x_train, y_train)
            model.fit(x_b, y_b)
            y_pred[:,i] = model.predict(x_test).ravel()

        error[degree] = np.mean(
            np.mean((y_test - y_pred)**2, axis=1, keepdims=True)
            )
        
        bias[degree] = np.mean(
            (y_test - np.mean(y_pred, axis=1, keepdims=True))**2
            )
        var[degree] = np.mean(np.var(y_pred, axis=1, keepdims=True))

    return error, bias, var


def plot_bias_variance(n_values):
    """
    Plot bias-variance decomposition for one or multiple dataset sizes.

    Parameters:
    - n_values : int or list of ints      Number of data points. Example:  40  or  [40, 100, 400]
    """

    # Convert a single integer into a list
    if isinstance(n_values, int):
        n_values = [n_values]

    n_plots = len(n_values)

    if n_plots == 1:
        fig, ax = plt.subplots(figsize=(6.6, 4.0))
        axes = [ax]

    else:
        fig, axes = plt.subplots(1, n_plots, figsize=(5 * n_plots, 4), sharey=True)

    for ax, n in zip(axes, n_values):
        error, bias, var = bias_variance(n=n)
        degrees = np.arange(len(error))

        ax.plot(degrees, error, "o-", color="midnightblue", label="Test error")
        ax.plot(degrees, bias, "s-", color="mediumvioletred", label=r"Bias$^2$")
        ax.plot(degrees, var, "d-", color="chocolate", label="Variance")

        ax.set_yscale("log")
        ax.set_xlabel("Polynomial degree")
        ax.set_title(f"n = {n} data points")
        ax.grid(alpha=0.3)

    axes[0].set_ylabel("MSE decomposition")
    axes[0].legend()

    plt.tight_layout()
    plt.show()
```

2) Discussion:

The model dominated by bias for low polynomial degrees, approximately $0\leq d \leq 3$. This is where the bias dominates the MSE, meaning that the model is too simple to reproduce the original function. This is underfitting.

The model is dominated by variance for higher polynomial degrees, approximately $d \geq 11$. For these degrees the variance become very high and it dominates the test error. This indicates that the model is overfitted.

For intermediate polynomial degrees, approximately $4\leq d \leq 11$, the bias is reduced and both the bias and the variance is relatively small. This indicates that these degrees are sufficient for the model, illustarting the bias-variance tradeoff. 

The optimal polynomial degree is naturally at near the minima of the test eroor, so around $d\approx 8$, where both the bias and the variance is relatively low.

We see that when we increase the number of data points the bias is roughly unchanged while the variance of the model reduces. For $n=40$ the variance increases notably for higher polynomial degrees, with $d\approx 8$ giving the best fit. For $n=100$ the higher degree models are more stable and the minima of the test error moves to higher polynomial degrees, with the optimal degree being around $d\approx 11$. For $n=400$ data points, the variance is substantially decreased compared to the previous cases and remains small across almost all polynomial degrees. The bias dominates the test error in this case. The minima is shifted to $d\approx 12$.

Consequently, increasing the amount of training data allows us to use more complex models without without being penalized by variance.

```python
# using Ridge regression at d = 12

def ridge_bias_variance(n, degree=12, n_bootstraps=100):

    np.random.seed(2026)
    x = np.linspace(-3, 3, n).reshape(-1, 1)
    y = np.exp(-x**2) + 1.5 * np.exp(-(x - 2)**2) + np.random.normal(0, 0.1, x.shape)

    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2026)

    lambdas = np.logspace(-6, 2, 9)
    error = np.zeros(len(lambdas))
    bias = np.zeros(len(lambdas))
    var = np.zeros(len(lambdas))

    for j, lmbda in enumerate(lambdas):
        model = make_pipeline(
            PolynomialFeatures(degree=degree),
            Ridge(alpha=lmbda, fit_intercept=False)
        )

        # one column for each bootstrap
        y_pred = np.zeros((len(y_test), n_bootstraps))

        # start resampling
        for i in range(n_bootstraps):
            x_b, y_b = resample(x_train, y_train)
            model.fit(x_b, y_b)
            y_pred[:,i] = model.predict(x_test).ravel()

        error[j] = np.mean(
            np.mean((y_test - y_pred)**2, axis=1, keepdims=True)
            )
        
        bias[j] = np.mean(
            (y_test - np.mean(y_pred, axis=1, keepdims=True))**2
            )
        var[j] = np.mean(np.var(y_pred, axis=1, keepdims=True))

    return lambdas, error, bias, var


# plotting
lambdas, error, bias, var = ridge_bias_variance(n=40)

plt.figure(figsize=(6.6, 4.0))
plt.plot(lambdas, error, "o-", label="Test error")
plt.plot(lambdas, bias, "s-", label=r"Bias$^2$")
plt.plot(lambdas, var, "d-", label="Variance")
plt.xscale("log")
plt.yscale("log")
plt.xlabel(r"$\lambda$")
plt.ylabel("MSE decomposition")
plt.title("Ridge regression, degree = 12")
plt.legend()
plt.show()
```

4) Discussion

For small regularization, aproximately $10^{-6} \leq \lambda \leq 10^{-3}$ the variance is very large and domantes the test error. As the regularization increases, the variance decreases because Ridge suppresses large coefficients. At degree 12, the model is relatively flexible, and we saw earlier that such a complex model can have high variance unless we use a larger training data set. 

For $\lambda \approx 10^{-2}$ the variance, bias and test error decreases substantially. For larger vlaues of $\lambda$, the model becomes increasingly regularized and the bias increases because coefficients are driven closer to zero, reducing the models flexibility. Thus, the regularization parameter $\lambda$ mainly attacks the varaince of the model, but the cost of this is an increase in bias for large regularization parameters.

From the figure, $\lambda \approx 1$ is the optimal regularization as the test error is at a minima and the variance is lowest here.

## 4.1 Solution

```python
def kfold_cv(X, y, model_degree, k=5, ridge = False, lam = 0):
    kf = KFold(n_splits=k, shuffle=True, random_state=2026)
    mse_list = []

    for train_idx, test_idx in kf.split(X):
        if not ridge:
            model = make_pipeline(PolynomialFeatures(model_degree), LinearRegression())
        else:
            model = make_pipeline(PolynomialFeatures(model_degree), Ridge(lam))
        model.fit(X[train_idx], y[train_idx])
        y_actual, y_pred = y[test_idx], model.predict(X[test_idx])
        mse_list.append(mean_squared_error(y_actual, y_pred))

    return np.mean(mse_list), np.std(mse_list)

        
```
